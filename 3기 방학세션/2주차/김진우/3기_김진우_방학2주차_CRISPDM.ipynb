###간단한 DF

```python
import pandas as pd
import numpy

customers = pd.DataFrame({
    "customer_id" : [1, 2, 3, 4],
    "name": ["A", "B", "C", "D"]})

orders = pd.DataFrame({
    "order_id" : [101, 102, 103, 104],
    "customer_id": [1, 1, 2, 5],
    "amount" : [30000, 12000, 45000, 9999]})

customers, orders
```

###INNER JOIN : 양쪽에 모두 있는 행만 합치기

```python
inner = pd.merge(customers, orders, on="customer_id", how="inner")
inner
```

###LEFT JOIN : 왼쪽 전부 살리기

```python
left = pd.merge(customers, orders, on="customer_id", how="left")
left
```

```python
left_only = left[left["order_id"].isnull()]
left_only
```

###RIGHT JOIN : 오른쪽 전부 살리기

```python
right = pd.merge(customers, orders, on="customer_id", how="right", indicator=True)
right
```

```python
right_only = right[right["name"].isnull()]
right_only
```

###FULL OUTER JOIN : 양쪽 모든 행 합치기

```python
outer = pd.merge(customers, orders, on="customer_id", how="outer", indicator=True)
outer
```

###CONCAT

```python
concat = pd.concat([customers, orders])
concat
```

```python
concat = pd.concat([customers, orders], axis = 1)
concat
```

###Olist data join

```python
import pandas as pd

#파일 불러오기
customers = pd.read_csv("olist_customers_dataset.csv")
geolocation = pd.read_csv("olist_geolocation_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
order_payments = pd.read_csv("olist_order_payments_dataset.csv")
order_reviews = pd.read_csv("olist_order_reviews_dataset.csv")
orders = pd.read_csv("olist_orders_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
product_category = pd.read_csv("product_category_name_translation.csv")

#조인하기
merged_df = customers.merge(orders, on='customer_id', how='left')
merged_df = merged_df.merge(order_items, on='order_id', how='left')
merged_df = merged_df.merge(products, on='product_id', how='left')
merged_df = merged_df.merge(sellers, on='seller_id', how='left')
merged_df = merged_df.merge(product_category, on='product_category_name', how='left')
merged_df = merged_df.merge(order_payments, on='order_id', how='left')
merged_df = merged_df.merge(order_reviews, on='order_id', how='left')

display(merged_df)
print(merged_df.columns)
```
