###간단한 DF

```python
import pandas as pd
import numpy

df1 = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["A", "B", "C", "D"]
})

df2 = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [1, 1, 2, 5],
    "amount": [30000, 12000, 45000, 9999]
})

df1, df2
```

###INNER JOIN : 양쪽에 모두 있는 행만 합치기

```python
inner = pd.merge(df1, df2, on="customer_id", how="inner")
inner
```

###LEFT JOIN : 왼쪽 전부 살리기

```python
left = pd.merge(df1, df2, on="customer_id", how="left")
left
```

```python
left_only = left[left["order_id"].isnull()]
left_only
```

###RIGHT JOIN : 오른쪽 전부 살리기

```python
right = pd.merge(df1, df2, on="customer_id", how="right")
right
```

```python
right_only = right[right["name"].isnull()]
right_only
```

###FULL OUTER JOIN : 양쪽 모든 행 합치기

```python
outer = pd.merge(df1, df2, on="customer_id", how="outer", indicator=True)
outer
```

###CONCAT

```python
concat = pd.concat([df1, df2])
concat
```

```python
concat = pd.concat([df1, df2], axis=1)
concat
```
