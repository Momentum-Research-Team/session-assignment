## 문제 2. 결과 해석해보기

다음 질문 각각에 답하시오.

1. 도출된 정확도 수치는 무엇을 의미하는가?
2. 모델이 질병이라고 예측한 사람은 총 몇 명인가?
3. 도출된 정밀도 수치는 무엇을 의미하는가?
4. 실제 질병 환자는 총 몇 명인가?
5. 도출된 재현율 수치는 무엇을 의미하는가?
6. 이 모델이 놓친 실제 환자는 몇 명인가?
7. 질병 환자를 놓치지 않는 것이 중요하다면 어떤 지표를 우선 확인해야 하는가?


## 문제 1 답안


````python
tp = 90
tn = 45
fp = 10
fn = 35

total = tp + tn + fp + fn

accuracy = (tp + tn) / total
precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1_score = 2 * precision * recall / (precision + recall)

print("TP:", tp)
print("TN:", tn)
print("FP:", fp)
print("FN:", fn)

print(f"Accuracy : {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall   : {recall:.2f}")
print(f"F1-score : {f1_score:.2f}")
````

2. 모델이 질병이라고 예측한 사람은 TP와 FP를 합한 100명이다.
3. 정밀도 0.90은 모델이 질병이라고 예측한 100명 중 실제 질병 환자가 90명, 즉 90%라는 의미이다.
4. 실제 질병 환자는 TP와 FN을 합한 125명이다.
5. 재현율 0.72는 실제 질병 환자 125명 중 90명, 즉 72%를 질병 환자로 찾아냈다는 의미이다.
6. 모델이 놓친 실제 환자는 FN에 해당하는 35명이다.
7. 질병 환자를 놓치지 않는 것이 중요하다면 재현율(Recall)을 우선 확인해야 한다.
