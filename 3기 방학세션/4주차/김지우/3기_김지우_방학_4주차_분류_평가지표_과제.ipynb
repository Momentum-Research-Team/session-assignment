## 문제 1 : 혼동행렬 파이썬 코드 작성

```python
tp = 90
tn = 45
fp = 10
fn = 35

total = tp + tn + fp + fn

accuracy = (tp + tn) / total
precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1_score = 2 * (precision * recall) / (precision + recall)

print("TP:", tp)
print("TN:", tn)
print("FP:", fp)
print("FN:", fn)

print(f"Accuracy : {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall   : {recall:.2f}")
print(f"F1-score : {f1_score:.2f}")
```
